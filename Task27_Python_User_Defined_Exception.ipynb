```python
# define Python user-defined exceptions
class InvalidAgeException(Exception):
    "Raised when the input value is less than 18"
    pass

# you need to guess this number
number = 18

try:
    input_num = int(input("Enter a number: "))
    if input_num < number:
        raise InvalidAgeException
    else:
        print("Eligible to Vote")

except InvalidAgeException:
    print("Exception occurred: Invalid Age")
```

```python
class SalaryNotInRangeError(Exception):
    """Exception raised when salary is outside the allowed range."""

    def __init__(self, salary, message="Salary must be between 5000 and 15000"):
        self.salary = salary
        self.message = message
        super().__init__(self.message)


try:
    salary = int(input("Enter salary amount: "))

    if not 5000 < salary < 15000:
        raise SalaryNotInRangeError(salary)

    print("Salary is within the allowed range.")
    print("Your salary is:", salary)

except ValueError:
    print("Error: Please enter a valid integer.")

except SalaryNotInRangeError as e:
    print("Error:", e)
    print("Invalid salary:", e.salary)
```
