## 1. Creating a List

```python
fruits = ["apple", "banana", "mango"]
print(fruits)

# Mixed data types
mixed = [10, "Python", 2.5, True]
print(mixed)

# Empty list
empty = []
print(empty)
```

```python
# Creating a list using list() function
word = "hello"
letters = list(word)
print(letters)

numbers = list(range(1, 6))
print(numbers)
```

## 2. Accessing List Items (Indexing)

```python
colors = ["red", "green", "blue"]

print("colors[0] =", colors[0])
print("colors[2] =", colors[2])
```

```python
# Negative indexing
colors = ["red", "green", "blue"]

print("colors[-1] =", colors[-1])
print("colors[-3] =", colors[-3])
```

```python
# Slicing
nums = [10, 20, 30, 40, 50]

print(nums[1:4])
print(nums[:3])
print(nums[2:])
```

## 3. Adding and Updating Items

```python
# Updating an item
basket = ["milk", "eggs", "bread"]
basket[1] = "butter"
print(basket)
```

```python
# append(): add item at the end
basket = ["milk", "eggs", "bread"]
basket.append("rice")
print(basket)
```

```python
# extend(): add multiple items from another list
basket = ["milk", "eggs", "bread"]
extra = ["tea", "sugar"]
basket.extend(extra)
print(basket)
```

```python
# insert(): add item at a specific position
basket = ["milk", "eggs", "bread"]
basket.insert(1, "honey")
print(basket)
```

## 4. Removing Items

```python
# remove(), pop() and clear()
items = ["pen", "book", "bag", "ruler"]

items.remove("bag")
print(items)

last = items.pop()
print(items)
print("Popped item:", last)

items.clear()
print(items)
```

```python
# del keyword
items = ["pen", "book", "bag", "ruler"]

del items[0]
print(items)

del items
# print(items)  # Uncomment to see NameError: name 'items' is not defined
```

## 5. Copying a List

```python
# Using = does not create a copy, both names point to the same list
a = [1, 2, 3]
b = a

a.append(4)

print("a =", a)
print("b =", b)
```

```python
# Using copy() creates an independent list
a = [1, 2, 3]
b = a.copy()

a.append(4)

print("a =", a)
print("b =", b)
```

## 6. Other List Methods

```python
nums = [5, 2, 8, 2, 9, 1]

print("index of 8:", nums.index(8))
print("count of 2:", nums.count(2))

nums.sort()
print("sorted:", nums)

nums.sort(reverse=True)
print("descending:", nums)

nums.reverse()
print("reversed:", nums)
```

## 7. len() Function

```python
subjects = ["Math", "Physics", "Chemistry", "English"]
print(len(subjects))
print(max([4, 9, 1]), min([4, 9, 1]), sum([4, 9, 1]))
```

## 8. Membership Test

```python
subjects = ["Math", "Physics", "Chemistry"]

print("Physics" in subjects)
print("Biology" in subjects)
print("Biology" not in subjects)
```

## 9. Iterating Through a List

```python
subjects = ["Math", "Physics", "Chemistry"]

for subject in subjects:
    print(subject)
```

```python
# Using enumerate() to get index and value
subjects = ["Math", "Physics", "Chemistry"]

for i, subject in enumerate(subjects):
    print(i, subject)
```
