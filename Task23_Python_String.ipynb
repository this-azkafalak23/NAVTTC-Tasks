## 1. Creating Strings

```python
# Single and double quotes are the same
s1 = 'Python'
s2 = "Python"

print(s1)
print(s2)
print(s1 == s2)
```

## 2. Multiline Strings

```python
# Triple quotes create a multiline string
poem = """Roses are red,
Violets are blue,
Python is fun,
And so are you."""

print(poem)
```

## 3. Accessing String Characters

```python
language = "Programming"

# Positive indexing
print(language[0])
print(language[3])
```

```python
# Negative indexing
language = "Programming"

print(language[-1])
print(language[-4])
```

```python
# IndexError when index is out of range
language = "Programming"

try:
    print(language[20])
except IndexError as e:
    print("IndexError:", e)
```

```python
# Slicing
language = "Programming"

print(language[0:4])
print(language[3:7])
print(language[:5])
print(language[5:])
print(language[::-1])
```

## 4. Strings are Immutable

```python
word = "Python"

try:
    word[0] = "J"
except TypeError as e:
    print("TypeError:", e)
```

```python
# Create a new string instead of modifying the old one
first = "Hello"
second = "World"

greeting = first + " " + second
print(greeting)

# Repetition
print("ha" * 3)
```

## 5. String Methods

```python
text = "I like Java."

new_text = text.replace("Java", "Python")
print(new_text)
print(text)  # original is unchanged
```

```python
msg = "  Hello Python World  "

print(msg.upper())
print(msg.lower())
print(msg.strip())
print(msg.title())
print(msg.split())
print(msg.strip().startswith("Hello"))
print(msg.strip().endswith("World"))
print(msg.find("Python"))
print(msg.count("o"))
```

```python
# join(): combine list items into one string
words = ["Learning", "Python", "is", "fun"]
sentence = " ".join(words)
print(sentence)
```

## 6. Membership Test

```python
print("Py" in "Python")
print("Java" in "Python")
print("Java" not in "Python")
```

## 7. Iterating Through a String

```python
name = "Ali"

for ch in name:
    print(ch)
```

## 8. String Length

```python
name = "Muhammad"
print(len(name))

print(len(""))
print(len("Hello World"))  # space is also counted
```

## 9. Escape Sequences

```python
# Escaping quotes with backslash
sentence1 = "She said, \"It's a nice day\""
sentence2 = 'She said, "It\'s a nice day"'

print(sentence1)
print(sentence2)
```

```python
# Common escape sequences
print("Line 1\nLine 2")
print("Name:\tAli")
print("Path: C:\\Users\\Ali")
print("Single quote: \'")
print("Double quote: \"")
```

## 10. String Formatting (f-Strings)

```python
name = "Sara"
age = 22

print(f"{name} is {age} years old.")
print(f"Next year {name} will be {age + 1}.")
```

```python
# Formatting numbers inside f-strings
pi = 3.14159265
price = 1234567

print(f"Pi rounded: {pi:.2f}")
print(f"Price: {price:,}")
```
