## 1. Creating a Set

```python
# Set of integers
roll_numbers = {101, 105, 103, 102}
print("Roll numbers:", roll_numbers)

# Set of strings
primary_colors = {"red", "blue", "yellow"}
print("Colors:", primary_colors)

# Set of mixed data types
mixed = {"Python", 42, 3.14, (1, 2)}
print("Mixed:", mixed)
```

```python
# A set cannot contain mutable items like lists
try:
    bad_set = {1, [2, 3]}
except TypeError as e:
    print("TypeError:", e)
```

## 2. Creating an Empty Set

```python
# {} creates an empty dictionary, use set() for an empty set
empty_set = set()
empty_dict = {}

print(type(empty_set))
print(type(empty_dict))
```

## 3. Duplicate Items

```python
# Sets automatically remove duplicates
values = {5, 3, 5, 9, 3, 1}
print(values)

# Removing duplicates from a list using set()
marks = [10, 20, 10, 30, 20, 40]
unique_marks = list(set(marks))
print(unique_marks)
```

## 4. Adding and Updating Items

```python
# add(): add a single item
cities = {"Lahore", "Karachi"}
print("Before:", cities)

cities.add("Islamabad")
print("After :", cities)
```

```python
# update(): add multiple items from a list, tuple or another set
brands = {"Nike", "Adidas"}
new_brands = ["Puma", "Nike", "Reebok"]

brands.update(new_brands)
print(brands)
```

```python
# Indexing is not supported because sets are unordered
items = {"a", "b", "c"}

try:
    print(items[0])
except TypeError as e:
    print("TypeError:", e)
```

## 5. Removing Items

```python
# discard(): removes item, no error if item is missing
subjects = {"Math", "Physics", "Chemistry"}

subjects.discard("Physics")
print(subjects)

subjects.discard("Biology")  # no error
print(subjects)
```

```python
# remove(): removes item, raises KeyError if item is missing
subjects = {"Math", "Physics", "Chemistry"}

subjects.remove("Math")
print(subjects)

try:
    subjects.remove("Biology")
except KeyError as e:
    print("KeyError:", e)
```

```python
# pop() removes a random item, clear() empties the set
nums = {1, 2, 3, 4}

removed = nums.pop()
print("Popped:", removed)
print("Remaining:", nums)

nums.clear()
print("After clear:", nums)
```

## 6. Built-in Functions with Set

```python
nums = {4, 8, 15, 16, 23, 42}

print("len   :", len(nums))
print("max   :", max(nums))
print("min   :", min(nums))
print("sum   :", sum(nums))
print("sorted:", sorted(nums))
print("all   :", all(nums))
print("any   :", any(nums))
```

```python
# enumerate()
for index, value in enumerate({"x", "y", "z"}):
    print(index, value)
```

## 7. Iterating Over a Set

```python
animals = {"cat", "dog", "horse"}

for animal in animals:
    print(animal)
```

## 8. Number of Set Elements

```python
odd_numbers = {1, 3, 5, 7, 9}
print("Set:", odd_numbers)
print("Total elements:", len(odd_numbers))
```

## 9. Membership Test

```python
fruits = {"apple", "banana", "mango"}

print("apple" in fruits)
print("grapes" in fruits)
print("grapes" not in fruits)
```

## 10. Set Operations

```python
# Union: all elements of both sets
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Using | :", A | B)
print("Using union():", A.union(B))
```

```python
# Intersection: common elements
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Using & :", A & B)
print("Using intersection():", A.intersection(B))
```

```python
# Difference: elements in A but not in B
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("A - B:", A - B)
print("B - A:", B - A)
print("Using difference():", A.difference(B))
```

```python
# Symmetric difference: elements in either set but not in both
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Using ^ :", A ^ B)
print("Using symmetric_difference():", A.symmetric_difference(B))
```

## 11. Checking if Two Sets are Equal

```python
A = {1, 2, 3}
B = {3, 2, 1}
C = {1, 2, 4}

print(A == B)
print(A == C)

if A == B:
    print("Set A and Set B are equal")
else:
    print("Set A and Set B are not equal")
```

## 12. Other Set Methods

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {7, 8}

print("A is subset of B   :", A.issubset(B))
print("B is superset of A :", B.issuperset(A))
print("A and C disjoint   :", A.isdisjoint(C))

D = A.copy()
D.add(99)
print("A:", A)
print("D (copy of A):", D)
```

```python
# In-place update methods
A = {1, 2, 3, 4}
B = {3, 4, 5}

A.difference_update(B)
print("difference_update:", A)

X = {1, 2, 3}
X.intersection_update({2, 3, 9})
print("intersection_update:", X)

Y = {1, 2, 3}
Y.symmetric_difference_update({3, 4})
print("symmetric_difference_update:", Y)
```
