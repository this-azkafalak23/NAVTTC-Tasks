## 1. Creating a Dictionary

```python
# Dictionary of key: value pairs
student = {
    "name": "Ali",
    "age": 21,
    "city": "Lahore"
}

print(student)
```

```python
# Creating a dictionary using dict()
student2 = dict(name="Sara", age=20, city="Karachi")
print(student2)

# From a list of tuples
student3 = dict([("name", "Bilal"), ("age", 22)])
print(student3)

# Empty dictionary
empty = {}
print(empty, type(empty))
```

## 2. Valid and Invalid Dictionaries

```python
# Integer keys
d1 = {1: "one", 2: "two", 3: "three"}
print(d1)

# Tuple as a key
d2 = {(1, 2): "point A", 3: "three"}
print(d2)

# String key with a list as value
d3 = {"Pakistan": ["Lahore", "Karachi", "Multan"]}
print(d3)
```

```python
# Invalid: a list cannot be used as a key
try:
    bad = {1: "Hello", [1, 2]: "Hi"}
except TypeError as e:
    print("TypeError:", e)
```

```python
# Keys must be unique: the later value overwrites the earlier one
marks = {
    "Ali": 70,
    "Sara": 85,
    "Ali": 90   # duplicate key
}

print(marks)
```

## 3. Accessing Dictionary Items

```python
capitals = {"Pakistan": "Islamabad", "India": "New Delhi", "China": "Beijing"}

print(capitals["Pakistan"])
print(capitals["China"])
```

```python
# get() returns None (or a default) instead of raising an error
capitals = {"Pakistan": "Islamabad", "India": "New Delhi"}

print(capitals.get("India"))
print(capitals.get("Japan"))
print(capitals.get("Japan", "Not found"))

try:
    print(capitals["Japan"])
except KeyError as e:
    print("KeyError:", e)
```

## 4. Adding Items

```python
capitals = {"Pakistan": "Islamabad", "India": "New Delhi"}

capitals["Japan"] = "Tokyo"
print(capitals)
```

## 5. Changing Items

```python
capitals = {"Pakistan": "Karachi", "India": "New Delhi"}

# Update the value of an existing key
capitals["Pakistan"] = "Islamabad"
print(capitals)
```

```python
# update(): add or change multiple items
capitals = {"Pakistan": "Islamabad"}

capitals.update({"India": "New Delhi", "Pakistan": "Islamabad (Capital)"})
print(capitals)
```

## 6. Removing Items

```python
# del statement
capitals = {"Pakistan": "Islamabad", "India": "New Delhi", "Japan": "Tokyo"}

del capitals["Japan"]
print(capitals)
```

```python
# pop() and popitem()
capitals = {"Pakistan": "Islamabad", "India": "New Delhi", "Japan": "Tokyo"}

removed = capitals.pop("India")
print("Removed:", removed)
print(capitals)

last = capitals.popitem()
print("Last item removed:", last)
print(capitals)
```

```python
# clear(): remove all items
capitals = {"Pakistan": "Islamabad", "India": "New Delhi"}

capitals.clear()
print(capitals)
```

## 7. Iterating Through a Dictionary

```python
capitals = {"Pakistan": "Islamabad", "India": "New Delhi", "Japan": "Tokyo"}

# Keys
for country in capitals:
    print(country)

print()

# Values
for country in capitals:
    print(capitals[country])
```

```python
# Using keys(), values() and items()
capitals = {"Pakistan": "Islamabad", "India": "New Delhi", "Japan": "Tokyo"}

print(capitals.keys())
print(capitals.values())
print(capitals.items())

for country, capital in capitals.items():
    print(f"The capital of {country} is {capital}")
```

## 8. Dictionary Length

```python
print(len({"a": 1, "b": 2}))
print(len({10: "ten", 20: "twenty", 30: "thirty"}))
print(len({}))
```

## 9. Dictionary Methods

```python
# copy(): creates an independent copy
original = {"x": 1, "y": 2}
duplicate = original.copy()

duplicate["z"] = 3
print("original :", original)
print("duplicate:", duplicate)
```

```python
# setdefault() and fromkeys()
d = {"a": 1}
d.setdefault("b", 2)    # adds key if it does not exist
d.setdefault("a", 100)  # existing key is unchanged
print(d)

keys = ["x", "y", "z"]
print(dict.fromkeys(keys, 0))
```

## 10. Membership Test

```python
file_types = {".txt": "Text File", ".pdf": "PDF Document", ".jpg": "JPEG Image"}

print(".pdf" in file_types)
print(".mp3" in file_types)
print(".mp3" not in file_types)

# in checks keys only, not values
print("PDF Document" in file_types)
print("PDF Document" in file_types.values())
```

## 11. Nested Dictionary

```python
students = {
    "s1": {"name": "Ali", "marks": 80},
    "s2": {"name": "Sara", "marks": 92}
}

print(students["s2"]["name"])
print(students["s1"]["marks"])

for sid, info in students.items():
    print(sid, info["name"], info["marks"])
```
